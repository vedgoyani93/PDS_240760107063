**Aim:** Create various types of plots using Matplotlib: line plot, bar chart, histogram, pie chart, scatter plot using subplots and figure properties.

### 1. Load Dataset

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('adult.csv')
df['income'] = df['income'].astype(str).str.strip().str.rstrip('.')
df.head(2)
```

### 2. Line Plot, Bar Chart, Histogram, Pie Chart, Scatter Plot using Subplots & Figure Properties

```python
# Prepare aggregated/sample data
age_hours = df.groupby('age')['hours-per-week'].mean()
top_edu = df['education'].astype(str).str.strip().value_counts().head(5)
income_counts = df['income'].value_counts()
sample_df = df.sample(300, random_state=42)

# Create 2x3 Subplots with Figure Properties
fig, axes = plt.subplots(2, 3, figsize=(14, 7))
fig.suptitle('Matplotlib Subplots: Various Plot Types', fontsize=14, fontweight='bold')

# 1. Line Plot
axes[0, 0].plot(age_hours.index, age_hours.values, color='blue')
axes[0, 0].set_title('1. Line Plot')
axes[0, 0].set_xlabel('Age')
axes[0, 0].set_ylabel('Hours/Week')
axes[0, 0].grid(True)

# 2. Bar Chart
axes[0, 1].bar(top_edu.index, top_edu.values, color='coral')
axes[0, 1].set_title('2. Bar Chart')
axes[0, 1].tick_params(axis='x', rotation=30)

# 3. Histogram
axes[0, 2].hist(df['age'], bins=20, color='lightgreen', edgecolor='black')
axes[0, 2].set_title('3. Histogram')
axes[0, 2].set_xlabel('Age')

# 4. Pie Chart
axes[1, 0].pie(income_counts, labels=income_counts.index, autopct='%1.1f%%', colors=['skyblue', 'salmon'])
axes[1, 0].set_title('4. Pie Chart')

# 5. Scatter Plot
axes[1, 1].scatter(sample_df['age'], sample_df['hours-per-week'], color='purple', alpha=0.5)
axes[1, 1].set_title('5. Scatter Plot')
axes[1, 1].set_xlabel('Age')
axes[1, 1].set_ylabel('Hours/Week')

# Remove unused 6th subplot
fig.delaxes(axes[1, 2])

plt.tight_layout()
plt.show()
```
