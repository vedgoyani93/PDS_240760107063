**Aim:** Apply data preprocessing techniques: slicing, filtering, sorting, concatenating, aggregating and normalizing data using Pandas.

### 1. Load Dataset

```python
import pandas as pd

df = pd.read_csv('adult.csv')
print('Dataset Shape:', df.shape)
df.head(2)
```

### 2. Slicing & Filtering

```python
# Slicing with .iloc and .loc
print('Positional Slicing (.iloc):\n', df.iloc[0:3, 0:4])
print('\nLabel Slicing (.loc):\n', df.loc[0:2, ['age', 'education', 'hours-per-week']])

# Filtering
filtered = df[(df['age'] > 50) & (df['hours-per-week'] >= 40)]
print(f'\nFiltered Count (Age > 50 & Hours >= 40): {len(filtered)}')
```

### 3. Sorting & Concatenating

```python
# Sorting
print('Sorted by Age (Descending):\n', df[['age', 'education', 'hours-per-week']].sort_values(by='age', ascending=False).head(3))

# Concatenating (Vertical and Horizontal)
v_concat = pd.concat([df.iloc[:2, :3], df.iloc[2:4, :3]], axis=0)
print('\nVertical Concatenation:\n', v_concat)

h_concat = pd.concat([df[['age']].head(3), df[['hours-per-week']].head(3)], axis=1)
print('\nHorizontal Concatenation:\n', h_concat)
```

### 4. Aggregating & Normalizing

```python
# Aggregating with groupby
print('Average Hours Worked by Education:\n', df.groupby('education')['hours-per-week'].mean().head(4))

# Normalizing Data (Min-Max Scaling)
df['age_norm'] = (df['age'] - df['age'].min()) / (df['age'].max() - df['age'].min())
print('\nOriginal vs Min-Max Normalized Age:\n', df[['age', 'age_norm']].head(3))
```
