**Aim:** Plot and analyze Normal and Exponential continuous probability distributions and demonstrate the Central Limit Theorem (Scenario 4: Delivery Times & Battery Durations).

### 1. Normal Distribution
Continuous probability distribution with parameters $\mu = 45$ (mean) and $\sigma = 6$ (standard deviation).

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm, expon

# Normal Distribution parameters
mu = 45
sigma = 6
x = np.linspace(mu - 4 * sigma, mu + 4 * sigma, 1000)
normal_pdf = norm.pdf(x, mu, sigma)

# Plot Normal Distribution
plt.figure(figsize=(8, 4))
plt.plot(x, normal_pdf, color="forestgreen")
plt.title(r"Normal Distribution ($\mu=45, \sigma=6$)")
plt.xlabel("X")
plt.ylabel("Probability Density")
plt.grid(True)
plt.show()
```

### 2. Exponential Distribution
Continuous probability distribution modeling arrival/waiting times with rate parameter $\lambda = 0.33$ (mean scale $= 1/\lambda = 3.0$).

```python
# Exponential Distribution parameters
lam = 0.33
x_exp = np.linspace(0, 18, 1000)
exponential_pdf = expon.pdf(x_exp, scale=3.0)

# Plot Exponential Distribution
plt.figure(figsize=(8, 4))
plt.plot(x_exp, exponential_pdf, color="darkmagenta")
plt.title(r"Exponential Distribution ($\lambda=0.33$)")
plt.xlabel("X")
plt.ylabel("Probability Density")
plt.grid(True)
plt.show()
```

### 3. Central Limit Theorem (CLT)
Demonstrating that the distribution of sample means approaches a Normal distribution as sample size increases, regardless of the underlying population distribution.

```python
np.random.seed(46)

# Generate an exponential population
population = np.random.exponential(scale=3.0, size=10000)

# Draw 1000 random samples of size 30 and calculate their means
sample_means = [np.mean(np.random.choice(population, size=30)) for _ in range(1000)]

# Plot Central Limit Theorem
plt.figure(figsize=(8, 4))
plt.hist(sample_means, bins=30, density=True, color="coral", edgecolor="black")
plt.title("Central Limit Theorem (Distribution of Sample Means)")
plt.xlabel("Sample Mean")
plt.ylabel("Density")
plt.grid(True)
plt.show()

print("Population mean:", round(np.mean(population), 2))
print("Mean of sample means:", round(np.mean(sample_means), 2))
```
