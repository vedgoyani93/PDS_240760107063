**Aim:** Use Pandas to create Series and DataFrames, perform data selection, filtering, grouping and aggregation operations.

### 1. Creating Series and DataFrame

```python
import pandas as pd

# Series
marks = pd.Series([85, 92, 78, 90, 88], index=["Maths", "Physics", "Chemistry", "English", "CS"])
print("Subject Marks Series:\n", marks)

# DataFrame (Student data)
student_data = {
    "RollNo": [101, 102, 103, 104, 105, 106],
    "Name": ["Rohan", "Sneha", "Vikas", "Ananya", "Manish", "Priya"],
    "Branch": ["CSE", "IT", "CSE", "ECE", "IT", "CSE"],
    "CGPA": [8.5, 9.1, 7.8, 8.2, 7.5, 8.9],
    "Semester": [5, 5, 5, 5, 5, 5]
}

df = pd.DataFrame(student_data)
print("\nStudent DataFrame:\n", df)
```

### 2. Data Selection

```python
# Selecting columns
print("Student Names:\n", df["Name"])
print("\nName and CGPA:\n", df[["Name", "CGPA"]])

# iloc selection
print("\nTop 3 students (iloc):\n", df.iloc[:3])
print("\nRow 2 details (iloc):\n", df.iloc[2, [1, 2, 3]])

# loc selection
print("\nRows 0 to 3 with Name & Branch (loc):\n", df.loc[0:3, ["Name", "Branch"]])
```

### 3. Data Filtering

```python
# Filter students with CGPA > 8.0
print("Students with CGPA > 8.0:\n", df[df["CGPA"] > 8.0])

# Filter CSE students with CGPA >= 8.5
print("\nCSE Students with CGPA >= 8.5:\n", df[(df["Branch"] == "CSE") & (df["CGPA"] >= 8.5)])
```

### 4. Grouping and Aggregation

```python
# Average CGPA by Branch
print("Average CGPA by Branch:\n", df.groupby("Branch")["CGPA"].mean())

# Summary stats per branch
print("\nBranch Summary Aggregation:\n", df.groupby("Branch").agg({
    "CGPA": ["mean", "max", "min"],
    "RollNo": "count"
}))
```
