**Aim:** Use SciPy's stats module to compute statistical measures and perform statistical tests on datasets.

### 1. Statistical Measures using SciPy

```python
from scipy import stats
import numpy as np

# Create Dataset (Student Examination Marks)
scores = [45, 55, 60, 60, 75, 80, 85, 90]
print("Dataset:", scores)

# Measures of Central Tendency (Mean, Median, Mode)
print("Mean               :", stats.tmean(scores))
print("Median             :", np.median(scores))
result = stats.mode(scores)
print(f"Mode               : {result.mode} (count: {result.count})")

# Measures of Dispersion & Shape (Std Dev, Variance, Skewness, Kurtosis)
print("Standard Deviation :", round(stats.tstd(scores), 4))
print("Variance           :", round(stats.tvar(scores), 4))
print("Skewness           :", round(stats.skew(scores), 4))
print("Kurtosis           :", round(stats.kurtosis(scores), 4))

# Overall SciPy Describe Summary
print("\nSciPy Describe Summary:\n", stats.describe(scores))
```

### 2. Probability Distributions (Normal Distribution)

```python
# Standard Normal Distribution: Probability Density Function (PDF) & Cumulative Distribution Function (CDF)
x = 0
print("Standard Normal PDF at x=0   :", round(stats.norm.pdf(x, loc=0, scale=1), 4))
print("Standard Normal CDF at Z=1.645:", round(stats.norm.cdf(1.645, loc=0, scale=1), 4))
```

### 3. Hypothesis Testing: One-Sample t-Test

```python
# Testing if average class score equals hypothesized benchmark (mu=65)
t_stat, p_val = stats.ttest_1samp(scores, popmean=65)

print("Sample:", scores)
print("t-statistic:", round(t_stat, 4))
print("p-value    :", round(p_val, 4))

if p_val < 0.05:
    print("Conclusion : Reject Null Hypothesis (Significant difference from benchmark).")
else:
    print("Conclusion : Fail to Reject Null Hypothesis (No significant difference from benchmark).")
```

### 4. Correlation Test (Pearson Correlation)

```python
# Pearson Correlation Test (Study Hours vs Exam Result)
study_hours = [2, 3, 5, 7, 9]
exam_result = [40, 52, 68, 80, 92]

result_corr = stats.pearsonr(study_hours, exam_result)
print("Correlation : ", round(result_corr.statistic, 4))
print("p-value     : ", round(result_corr.pvalue, 6))
```
