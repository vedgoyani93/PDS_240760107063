```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom, poisson

# Problem:
# A basketball player shoots 12 free throws with a 50% success rate.
# Find the probability of making exactly 6 shots.
# Then calculate and visualize the Binomial distribution.
n = 12
p = 0.5

probability = binom.pmf(6, n, p)

print("Probability of getting exactly 6 successful shots:", probability)

# Values of X
x = np.arange(0, 13)

# Binomial probabilities
y = binom.pmf(x, n, p)

print("Binomial Distribution")
for i in range(len(x)):
    print("Number of Successful Shots =", x[i], "Probability =", round(y[i], 4))
```

```python
# Problem:
# A city department records an average of 2 road maintenance calls per day.
# Find the probability of receiving exactly 1 call in a day.
# Then calculate and visualize the Poisson distribution.
lam = 2

probability = poisson.pmf(1, lam)

print("Probability of getting exactly 1 call:", probability)

# Values of X
x = np.arange(0, 9)

# Poisson probabilities
y = poisson.pmf(x, lam)
print("Poisson Distribution")
for i in range(len(x)):
    print("Number of Calls =", x[i], "Probability =", round(y[i], 4))
```

```python
plt.figure(figsize=(10, 4))

# Binomial plot
plt.subplot(1, 2, 1)
plt.bar(np.arange(0, 13), binom.pmf(np.arange(0, 13), 12, 0.5), color="darkorange")
plt.title("Binomial Distribution")
plt.xlabel("Number of Successful Shots")
plt.ylabel("Probability")

# Poisson plot
plt.subplot(1, 2, 2)
plt.bar(np.arange(0, 9), poisson.pmf(np.arange(0, 9), 2), color="dodgerblue")
plt.title("Poisson Distribution")
plt.xlabel("Number of Maintenance Calls")
plt.ylabel("Probability")

plt.tight_layout()
plt.show()
```
